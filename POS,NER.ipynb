```python
text = """
Sundar Pichai, the CEO of Google, visited Bengaluru on Monday.
He met several engineers at the Google office and discussed new
artificial intelligence projects. Microsoft CEO Satya Nadella also
visited India in 2025 and announced a $3 billion investment.
"""
```

```python
import nltk
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
sentence = sent_tokenize(text)
```

```python
all= []
for i in sentence:
    words = word_tokenize(i)
    words = [word for word in words if word not in stopwords.words('english')]
    pos_tags = nltk.pos_tag(words)
    all.append(pos_tags)
all
```

```python
#NER
all= []
for i in sentence:
    words = word_tokenize(i)
    words = [word for word in words if word not in stopwords.words('english')]
    pos_tags = nltk.pos_tag(words)
    Ne = nltk.ne_chunk(pos_tags).draw()
```
