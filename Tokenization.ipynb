```python
import nltk

nltk.download('punkt_tab')
```

```python
corpus = """
Dr. Sharma said, "I can't believe it's already 2026!"
He paid $3.88 for coffee, but the bill was actually $5.00.
Wow... that's expensive, isn't it?

My friend Rahul lives in Bengaluru. He works at OpenAI.
Can you believe that? I don't think so.

The students were learning NLP, AI, and Machine Learning.
They're studying tokenization because it's an important part of NLP.
"""
```

```python
#SENTENCE TOKENIZATION
from nltk.tokenize import sent_tokenize
documents = sent_tokenize(corpus)
documents
```

```python
type(documents)
```

```python
for i in documents:
    print(i)
```

```python
#WORD TOKENIZATION
from nltk.tokenize import word_tokenize
words = word_tokenize(corpus)
words
```

```python
type(words)
```

```python
for i in words:
    print(i)
```

```python
#WORD TOKENIZATION WITH PUNCTUATION
from nltk.tokenize import wordpunct_tokenize
words_with_punctuation = wordpunct_tokenize(corpus)
words_with_punctuation
```

```python
type(words_with_punctuation)
```

```python
for i in words_with_punctuation:
    print(i)
```

```python
#Treebank Tokenization
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
