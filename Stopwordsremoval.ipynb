```python
import nltk
nltk.download('stopwords')
```

```python
corpus ="The children were happily playing in the beautiful gardens, and they were running around while their parents watched them. One child was studying different subjects and another child was writing interesting stories. " \
"The people enjoyed these activities because they were learning new things every day."
```

```python
from nltk.tokenize import sent_tokenize,word_tokenize
from nltk.corpus import stopwords
sentences = sent_tokenize(corpus)
```

```python
from nltk.stem import PorterStemmer, SnowballStemmer
ps = PorterStemmer()
sn= SnowballStemmer(language='english')
```

```python
#stemming with PorterStemmer
processed_sentences = []
for i in sentences:
    words = word_tokenize(i)
    words = [ps.stem(word) for word in words if word not in stopwords.words('english')] 
    i = ' '.join(words)
    processed_sentences.append(i)
print(processed_sentences)
```

```python
#stemming with SnowballStemmer
processed_sentences = []
for i in sentences:
    words = word_tokenize(i)
    words = [sn.stem(word) for word in words if word not in stopwords.words('english')] 
    i = ' '.join(words)
    processed_sentences.append(i)
print(processed_sentences)
```

```python
#Lemmatization
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
processed_sentences = []
for i in sentences:
    words = word_tokenize(i)
    words = [lemmatizer.lemmatize(word,pos='v').lower() for word in words if word not in stopwords.words('english')] 
    i = ' '.join(words)
    processed_sentences.append(i)
print(processed_sentences)
```
