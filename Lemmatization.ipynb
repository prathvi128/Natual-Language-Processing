```python
words = [
    "studies",
    "studying",
    "running",
    "played",
    "cars",
    "mice",
    "children",
    "better",
    "geese",
    "feet"
]
```

```python
import nltk
nltk.download('wordnet')
```

```python
#Lemmatization
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()
```

```python
for word in words:
    wordnet_lemma = wordnet_lemmatizer.lemmatize(word,pos='n')
    print(wordnet_lemma)
```

```python
for word in words:
    wordnet_lemma = wordnet_lemmatizer.lemmatize(word,pos='v')
    print(wordnet_lemma)
```
