```python
words = [
    "playing",
    "played",
    "studies",
    "relational",
    "conditional",
    "generously",
    "organization",
    "organizational",
    "happiness",
    "running"
]
```

```python
#Porter Stemmer
from nltk.stem import PorterStemmer
ps = PorterStemmer()

for word in words:
    ps_stem = ps.stem(word)
    print(ps_stem)
```

```python
#Regexp Stemmer
from nltk.stem import RegexpStemmer
rs = RegexpStemmer('ing$|s$|ed$|able$', min=4)

for word in words:
    rs_stem = rs.stem(word)
    print(rs_stem)
```

```python
#Snowball Stemmer
from nltk.stem import SnowballStemmer

ss = SnowballStemmer(language='english')

for word in words:
    ss_stem = ss.stem(word)
    print(ss_stem)
```
